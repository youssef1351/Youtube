```python
# =========================================================
# Digital Egypt Cubs Initiative - Level 2 Final Project
# Task 1.1: Cashier Program for Future Mall
# =========================================================

# Step 1: Define the two parallel lists (5 items and 5 matching prices in EGP)
product_names = ["Milk", "Bread", "Coffee", "Cheese", "Olive Oil"]
product_prices = [35.0, 15.0, 120.0, 85.0, 250.0]

print("--- Welcome to Future Mall Cashier System ---")
print("Available Products:")
for i in range(len(product_names)):
    print(f"{i+1}. {product_names[i]} - {product_prices[i]} EGP")
print("---------------------------------------------")

total_price = 0.0

# Step 2: The purchase loop
while True:
    choice = input("\nEnter item number to buy (1-5), or type 'done' to finish: ").strip().lower()

    if choice == 'done':
        break

    if choice.isdigit():
        index = int(choice) - 1
        if 0 <= index < len(product_names):
            selected_item = product_names[index]
            item_price = product_prices[index]

            # Accumulate running total
            total_price += item_price
            print(f"Added {selected_item} ({item_price} EGP). Running Total: {total_price:.2f} EGP")
        else:
            print("Invalid item number. Please select between 1 and 5.")
    else:
        print("Invalid input! Enter a number (1-5) or 'done'.")

# Step 3: The discount condition (strictly exceeds 500 EGP)
discount_applied = False
discount_amount = 0.0

if total_price > 500:
    discount_amount = total_price * 0.10  # 10% discount
    final_total = total_price - discount_amount
    discount_applied = True
else:
    final_total = total_price

# Step 4: The Receipt Output
print("\n" + "="*35)
print("          FUTURE MALL RECEIPT          ")
print("="*35)
print(f"Subtotal:         {total_price:.2f} EGP")

if discount_applied:
    print(f"Discount (10%):  -{discount_amount:.2f} EGP")

print(f"Final Total:      {final_total:.2f} EGP")
print("-" * 35)
print("Thank you for shopping at Future Mall!")
print("="*35)
```
